```python
# collections
# 1. list
variable=[1, None, 2.4, 1, 2.4, 'Hello', True, 10+5j]
print (variable)
print (type(variable))
# 2. tuple
variable=(1, None, 2.4, 1, 2.4, 'Hello', True, 10+5j)
print (variable)
print (type(variable))
# 3. set
variable={1, None, 2.4, 1, 2.4, 'Hello', True, 10+5j}
print (variable)
print (type(variable))
# 4. dictionary
variable={10: "ten", 20: "twenty", 30: "thirty", 40: "forty"}
print (variable)
print (type(variable))
```
