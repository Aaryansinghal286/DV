# Numpy -> Numerical Python

1. Numpy is one of the most important foundational package for numerical computing in python.
2. It is an open source python package, that is widely used in science and engineering.
3. The Numpy library contains multi-dimensional array data structure, such as homogenous , n-dimensional (ND) array and a large library of function that operates efficiently on these data structures.

# Why do we use python Numpy?
* We use python Numpy array, instead of list beacuse of the below three reasons:-
  > - less memory
  > - fast
  > - convenient to undertand.
  > - vectorization (way faster than loop in list).

# Why not Python in-built data structure list ?
* A python list stores elements + meta data + pointers -> thus using more memory.
* A Numpy array stores data in a continuous block -> more efficeint and compact.


# Difference between List and Array
* List is an in-built data structure whereas Numpy is a python Library.
* List is slower. Array is faster.
* List store heterogenous data, Array stores Homogenous data structure.
* Numpy library contains multi-dimensional array data structure. Whereas List conatain singke dimensional data structure.
