# compytation on numpy arrays
* > computation of numpy arrays can be very fast or it can be very slow.
* > The key to making it very fast it to use

```python
import numpy as np
x=np.array([1,2,3,4])
y=np.array([5,6,7,8])
xy=np.add(x,y)
print(xy)
```

```python
x=np.array([1,2,3,4])
y=np.array([5,6,7,8])
xy=np.subtract(x,y)
print(xy)
```

```python
x=np.array([1,2,3,4])
y=np.array([5,6,7,8])
xy=np.multiply(x,y)
print(xy)
```

```python
x=np.array([1,2,3,4])
y=np.array([5,6,7,8])
xy=np.divide(x,y)
print(xy)
```

```python
x=np.array([1,-2,3,-4,-2,-3,2,7,-8])
xy=np.absolute(x)
print(xy)
```

```python
x=np.array([1,-2,3,-4,-2,-3,2,7,-8])
np.exp(x)
np.exp2(x)
np.log(x)
np.log2(x)
```

# Advanced Ufunc features
* > 1. Specifying calculation

```python
import numpy as np
x=np.arange(2,20,2)
print(x)
z=np.empty((9))
print(z)
print(np.multiply(x,10,out=z))
```

* > 2. Aggregation (Reduce method)
    > - If you like to reduce an array with a particular operation , we can use the reduce method of any ufunc.
    > - A reduce repeatedly applies a given operation to the elements of an array until only a single result remains.

```python
x=np.arange(1,11)
print(x)
np.add.reduce(x)
```

```python
np.add.accumulate(x)
```
