# Labels with units

```python
hours=[1,2,3,4,5,6]
marks=[42,51,60,79,86,87]

fig,ax=plt.subplots(figsize=(6,3.5))
ax.plot(hours,marks,color='yellow', marker='s')
ax.set_title('First Matplotlib plot')
ax.set_xlabel('Observation')
ax.set_ylabel('Value')
plt.show()
```

# Controlling figure size

```python
x=np.arange(1,7)
y=x**2
fig, ax=plt.subplots(figsize=(8,3))
ax.plot(x,y, marker='o')
ax.set(title='wide figure', xlabel='x', ylabel='x squared')
plt.show()
```

```python
semesters = np.arange(1,7)
cgpa=[7.1, 7.4, 7.6, 8.0, 8.2, 8.5]
fig, ax=plt.subplots(figsize=(6,3.5))
ax.plot(semesters, cgpa, marker='o')
ax.set_xticks(semesters)
ax.set_yticks(np.arange(6.5, 9.1, 0.5))
ax.grid(axis='y', linestyle='--', alpha=0.45)
ax.set(title='semester wise CGPA', xlabel='Semester', ylabel='CGPA')
plt.show()
```
