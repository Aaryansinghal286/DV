The default value of separator (sep) is space "_" .

```python
a=10
b=15
c=6
print(a, b, c, sep="@")
print("hi")
```

the default value of end function: " " space

```python
a=10
b=15
c=6
print(a, b, c, sep="@", end="$")
print("hi")
```
