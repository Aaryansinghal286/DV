```python
a=10  #int
b=3 #float
c=int(a/b)
print(a, type(a))
print(b, type(b))
print(c, type(c))
```

```python
age=25
print(type(age))

# type conversion
age_str=str(age)
print(age_str)
print(type(age_str))
```

```python
x="123"
print(x, type(x))

y=float(x)
print(y, type(y))
```

```python
age=input("Enter your age: ")
print("Your age is:", age, type(age))
```

```python
# Simple caluclator
num1=int(input("Enter first number: "))
num2=int(input("Enter second number: "))
sum=num1+num2
difference=num1-num2
product=float(num1*num2)
division=float(num1/num2)
print("Sum:", sum, type(sum))
print("Difference:", difference, type(difference))
print("Product:", product, type(product))
print("Division:", division, type(division))
```

```python
a=5
b=10
add_result=a+b
sub_result=a-b
mult_result=a*b
div_result=a/b
floor_div_result=a//b
modulus_result=a%b
exponent_result=a**b
print(add_result)
print(sub_result)
print(mult_result)
print(div_result)
print(floor_div_result)
print(modulus_result)
print(exponent_result)
```
