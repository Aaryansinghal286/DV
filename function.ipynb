# Arbitrary Argument
    -> simple arbitrary (*args)

```python
def addition (*args):    #args is a tuple
    print(type(args))
    sum=0
    for i in args:
        sum+=i
    return sum
result=addition(6,9,4,7,3,2,5,0)
print(result)
```

-> Keyword arbitrary (**kwargs)

```python
def addition (**kwargs):    #kwargs is a dictionary
    print(type(kwargs))
    sum=0
    for i in kwargs:
        sum+=kwargs[i]      #sum = sum + kwargs[i]
    return sum
result=addition(a=6,b=9,c=4,d=7,e=3,f=2,g=5,h=0)
print(result)
```

```python
def addition (**kwargs):    #kwargs is a dictionary
    print(type(kwargs))
    sum=0
    for i in kwargs.keys():
        print(i)
    for i in kwargs.values():
        print(i)
    for i in kwargs.items():
        print(i)
    return sum
result=addition(a=6,b=9,c=4,d=7,e=3,f=2,g=5,h=0)
print(result)
```

# Q). WAP using function to generate perfect numbers from 1 to 1000.

```python
def is_perfect(n):
    sum_of_divisors = 0
    for i in range(1, n):
        if n % i == 0:
            sum_of_divisors += i
    return sum_of_divisors == n
for num in range(1, 1001):
    if is_perfect(num):
        print (num)
```

# Q). password Strength Checker

```python
# WAP for Password Strength Checker.
def is_strong_pswd(pswd):
    if len(pswd) < 8:
        return False
    if not any(char.isdigit() for char in pswd):
        return False
    if not any(char.isupper() for char in pswd):
        return False
    if not any(char.islower() for char in pswd):
        return False
    if not any(char in '!@#$%^&*()-_=+' for char in pswd):
        return False
    return True
print(is_strong_pswd("Weakpwd"))
print(is_strong_pswd("Str0ngPwd!"))
```

# Q). WAP to calculate total cost of items in the cart

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['quantity']
    return total_cost

cart=[
    {'name': 'Apple', 'price': 0.5, 'quantity': 4},
    {'name': 'Banana', 'price': 0.3, 'quantity': 6},
    {'name': 'Orange', 'price': 0.7, 'quantity': 3}
]
total_cost = calculate_total_cost(cart)
print(total_cost)
```

# Q). Check if a string is Palindrome

```python
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]
print(is_palindrome("A man a plan a canal Panama"))
print(is_palindrome("Hello"))
print(is_palindrome("Naman"))
```

# Q). WAP to find factorial using recursion.

```python
# WAP to find factorial using recursion.
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
n=int(input("Enter a number to find its factorial: "))
result = factorial(n)
print(f"The factorial of {n} is {result}")
```

# Q). WAP to find sum of n numbers using Lambda function

```python
# WAP to find sum of n natural numbers using recursion and lambda function.
def sum_natural(n):
    if n == 1:
        return 1
    return n + sum_natural(n - 1)

sum_natural_lambda = lambda n: sum_natural(n)
print(sum_natural_lambda(5))  # Example usage: sum of first 5 natural numbers
```
